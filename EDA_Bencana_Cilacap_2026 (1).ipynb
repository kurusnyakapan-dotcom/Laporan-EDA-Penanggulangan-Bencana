## Bab 1: Pendahuluan (Introduction)

### 1.1 Latar Belakang
Kabupaten Cilacap merupakan wilayah terluas di Provinsi Jawa Tengah dengan kondisi geografis dan topografi yang sangat beragam, mulai dari kawasan pesisir selatan, dataran rendah, hingga pegunungan dan perbukitan di bagian utara serta barat. Karakteristik geografis tersebut menjadikan Kabupaten Cilacap memiliki tingkat kerawanan yang tinggi terhadap berbagai jenis ancaman bencana hidrometeorologi maupun geologi, seperti cuaca ekstrem, tanah longsor, kekeringan, serta kebakaran hutan dan lahan.

Untuk mendukung upaya mitigasi, kesiapsiagaan, serta perencanaan alokasi anggaran penanggulangan bencana berbasis bukti (*evidence-based policy*), diperlukan pemahaman mendalam terhadap pola kejadian bencana yang tercatat secara historis. Melalui pendekatan *Exploratory Data Analysis* (EDA), data rekapitulasi kejadian bencana dan dampaknya dapat dieksplorasi secara sistematis guna mengungkap jenis bencana yang paling sering terjadi, bencana dengan dampak korban jiwa terbesar, serta bencana yang menimbulkan kerugian ekonomi paling signifikan.

### 1.2 Rumusan Masalah
Berdasarkan latar belakang di atas, analisis dalam laporan ini difokuskan untuk menjawab beberapa pertanyaan utama berikut:
1. Bagaimana kondisi kualitas data awal pada dataset rekapitulasi bencana Kabupaten Cilacap periode Januari – Agustus 2026 (terkait kelengkapan data, tipe data, maupun duplikasi)?
2. Bagaimana distribusi statistik dan karakteristik variabel utama kebencanaan, seperti frekuensi kejadian, jumlah rumah rusak, warga terdampak, dan total kerugian materiil?
3. Jenis bencana apa yang memiliki frekuensi kejadian tertinggi, dampak sosial (warga terdampak) terbesar, serta kerugian ekonomi paling tinggi di Kabupaten Cilacap?
4. Apakah terdapat nilai pencilan (*outliers*) pada dataset yang mencerminkan kejadian bencana berskala ekstrem dan membutuhkan perhatian khusus?

### 1.3 Tujuan Analisis
Tujuan dari pelaksanaan *Exploratory Data Analysis* (EDA) pada studi kasus ini adalah:
1. Melakukan inspeksi, pembersihan data (*data cleaning*), serta penanganan *missing values* dan data duplikat agar dataset siap dianalisis secara kuantitatif.
2. Menghitung ukuran pemusatan, penyebaran, kemencengan (*skewness*), dan keruncingan (*kurtosis*) untuk memahami profil distribusi dampak bencana.
3. Menerapkan teknik transformasi data dan rekayasa fitur (*feature engineering*) untuk menghasilkan indikator turunan yang lebih informatif, seperti rasio kerugian rata-rata per kejadian dan total kerusakan rumah.
4. Menyajikan visualisasi data serta mengidentifikasi pencilan (*outliers*) guna memberikan wawasan strategis bagi pengambilan keputusan mitigasi bencana di Kabupaten Cilacap.

## Bab 2: Pemahaman Data & Metodologi EDA (Data Understanding & Methodology)

### 2.1 Deskripsi Dataset
Dataset yang digunakan dalam analisis ini adalah **Rekapitulasi Kejadian dan Dampak Bencana di Kabupaten Cilacap** untuk periode **Januari hingga Agustus 2026**. Dataset awal memiliki dimensi **15 baris dan 39 kolom**, di mana setiap baris merepresentasikan kategori jenis bencana yang dipantau, sedangkan kolom-kolomnya memuat informasi kuantitatif mengenai cakupan wilayah terdampak, korban jiwa, kerusakan bangunan/infrastruktur, jumlah warga terdampak, pengungsi, serta taksiran kerugian materiil.

Secara garis besar, 39 atribut dalam dataset ini dikelompokkan ke dalam 6 kategori utama:
* **Identitas & Cakupan Wilayah (4 kolom):** `Jenis_Bencana`, `Kejadian`, `Kecamatan_Terdampak`, dan `Desa_Kelurahan_Terdampak`.
* **Korban Manusia (3 kolom):** `Meninggal_Dunia`, `Luka_Berat`, dan `Luka_Ringan`.
* **Kerusakan Pemukiman (6 kolom):** `Rumah_Roboh_Musnah`, `Rumah_Rusak_Berat`, `Rumah_Rusak_Sedang`, `Rumah_Rusak_Ringan`, `Rumah_Terendam_KK`, dan `Rumah_Terancam`.
* **Kerusakan Infrastruktur & Fasilitas Umum (17 kolom):** Mencakup sektor pendidikan (`Infra_Sekolah`), jaringan jalan (`Infra_Jln_Nas`, `Infra_Jln_Prov`, `Infra_Jln_Kab`, `Infra_Jln_Desa`), lingkungan/lahan (`Infra_Tebing`, `Infra_Sawah`, `Infra_Turap_Talud`, `Infra_Tanggul`, `Infra_Saluran_Air`, `Infra_Lahan_Terbakar`), ekonomi (`Infra_Pabrik`, `Infra_Pasar`, `Infra_Tempat_Usaha`), serta fasilitas publik lainnya (`Infra_Jembatan`, `Infra_Tempat_Ibadah`, `Infra_Fasum_Lainnya`).
* **Populasi Terdampak & Pengungsi (8 kolom):** Rincian warga terdampak dan pengungsi berdasarkan satuan jiwa, Kepala Keluarga (KK), laki-laki (`L`), dan perempuan (`P`).
* **Dampak Ekonomi (1 kolom):** `Jumlah_Kerugian` dalam satuan mata uang Rupiah (Rp).

### 2.2 Metodologi Analisis
Proses *Exploratory Data Analysis* (EDA) pada notebook ini dilaksanakan secara bertahap mengikuti alur kerja standar *Data Science*:
1. **Memuat & Memeriksa Data (Bab 3):** Mengimpor *library* komputasi numerik dan visualisasi (`pandas`, `numpy`, `matplotlib`, `seaborn`, `scikit-learn`), memuat dataset, serta memeriksa dimensi dan tipe data awal.
2. **Pembersihan Data (Bab 4):** Mengidentifikasi karakter tanda hubung (`-`) sebagai representasi nilai nol/kosong, menghapus baris duplikat yang tidak valid, membersihkan format mata uang (`Rp` dan tanda titik), serta mengonversi seluruh variabel kuantitatif menjadi tipe numerik.
3. **Eksplorasi Karakteristik Data (Bab 5):** Menghitung statistik deskriptif (rata-rata, standar deviasi, kuartil, nilai minimum dan maksimum) serta mengevaluasi tingkat asimetri distribusi melalui nilai *skewness* dan *kurtosis*.
4. **Transformasi & Rekayasa Fitur (Bab 6):** Menerapkan *Log Transformation* ($\log(1+x)$) untuk meredam kemencengan distribusi, normalisasi *Min-Max* untuk menyetarakan skala antarvariabel, serta membentuk fitur baru (`Kerugian_Per_Kejadian` dan `Total_Rumah_Rusak`).
5. **Visualisasi & Deteksi Pencilan (Bab 7 & 8):** Membangun grafik batang (*bar chart*) untuk membandingkan frekuensi dan kerugian antarjenis bencana, serta mengidentifikasi anomali/pencilan ekstrem pada variabel dampak bencana.
6. **Kesimpulan & Rekomendasi:** Merangkum temuan utama analisis dan merumuskan saran strategis berbasis data.

## Bab 4: Penanganan Data yang Hilang & Duplikat (Handling Missing Data)

```python
# 4.1 Identifikasi Nilai Kosong ('-')
df = df_raw.replace('-', np.nan)
print('Jumlah Missing Values per Kolom Utama:')
print(df[['Jenis_Bencana', 'Kejadian', 'Warga_Terdampak_Jiwa', 'Jumlah_Kerugian']].isnull().sum())

# Cek duplikasi nama bencana
print('\nDuplikasi Jenis Bencana:', df['Jenis_Bencana'].value_counts()[df['Jenis_Bencana'].value_counts() > 1])

# 4.3 Strategi Eliminasi: Hapus baris duplikat KEKERINGAN yang kosong (index 6)
df = df.drop(index=6).reset_index(drop=True)

# 4.4 Strategi Imputasi: Bersihkan format Rp & konversi NaN menjadi 0
df['Jumlah_Kerugian'] = (
    df['Jumlah_Kerugian'].astype(str)
    .str.replace('Rp', '', regex=False)
    .str.replace('.', '', regex=False)
    .astype(float)
)

num_cols = [c for c in df.columns if c != 'Jenis_Bencana']
for col in num_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0)

print('\nDimensi Dataset Setelah Pembersihan:', df.shape)
display(df[['Jenis_Bencana', 'Kejadian', 'Rumah_Rusak_Ringan', 'Warga_Terdampak_Jiwa', 'Jumlah_Kerugian']])
```

## Bab 5: Eksplorasi Karakteristik Data (Exploring Data Characteristics)

```python
# 5.1 Statistik Deskriptif Variabel Utama
main_vars = ['Kejadian', 'Rumah_Rusak_Ringan', 'Warga_Terdampak_Jiwa', 'Jumlah_Kerugian']
display(df[main_vars].describe())

# 5.2 Analisis Skewness & Kurtosis
skew_kurt = pd.DataFrame({
    'Skewness': df[main_vars].skew(),
    'Kurtosis': df[main_vars].kurt()
})
print('\nNilai Skewness dan Kurtosis:')
display(skew_kurt)
```

## Bab 6: Transformasi Data & Feature Engineering

```python
# 6.1 Log Transformation untuk memperbaiki skewness
df['Log_Kerugian'] = np.log1p(df['Jumlah_Kerugian'])
df['Log_Warga_Terdampak'] = np.log1p(df['Warga_Terdampak_Jiwa'])

# 6.2 Normalisasi MinMaxScaler (0 - 1)
scaler = MinMaxScaler()
df_scaled = pd.DataFrame(scaler.fit_transform(df[main_vars]), columns=[c + '_Scaled' for c in main_vars])

# 6.4 Feature Engineering: Kerugian Rata-rata per Kejadian & Total Rumah Rusak
df['Kerugian_Per_Kejadian'] = np.where(df['Kejadian'] > 0, df['Jumlah_Kerugian'] / df['Kejadian'], 0)
df['Total_Rumah_Rusak'] = (
    df['Rumah_Roboh_Musnah'] + df['Rumah_Rusak_Berat'] +
    df['Rumah_Rusak_Sedang'] + df['Rumah_Rusak_Ringan']
)

display(df[df['Kejadian'] > 0][['Jenis_Bencana', 'Kejadian', 'Total_Rumah_Rusak', 'Jumlah_Kerugian', 'Kerugian_Per_Kejadian']])
```

## Bab 7 & 8: Visualisasi Hubungan Data & Deteksi Pencilan (Outliers)

```python
# 7.1 Bar Chart Frekuensi Kejadian Bencana
plt.figure(figsize=(10, 5))
df_plot = df[df['Kejadian'] > 0].sort_values('Kejadian', ascending=True)
sns.barplot(data=df_plot, x='Kejadian', y='Jenis_Bencana', palette='viridis')
plt.title('Frekuensi Kejadian Bencana di Kabupaten Cilacap (Jan - Agu 2026)', fontweight='bold')
plt.xlabel('Jumlah Kejadian')
plt.ylabel('Jenis Bencana')
for idx, val in enumerate(df_plot['Kejadian']):
    plt.text(val + 1, idx, f'{int(val)}', va='center')
plt.tight_layout()
plt.show()

# 7.1 Bar Chart Kerugian Materiil
plt.figure(figsize=(10, 5))
df_rugi = df[df['Jumlah_Kerugian'] > 0].copy()
df_rugi['Kerugian_Juta'] = df_rugi['Jumlah_Kerugian'] / 1e6
df_rugi = df_rugi.sort_values('Kerugian_Juta', ascending=True)
sns.barplot(data=df_rugi, x='Kerugian_Juta', y='Jenis_Bencana', palette='magma')
plt.title('Total Kerugian Materiil per Jenis Bencana (Juta Rupiah)', fontweight='bold')
plt.xlabel('Kerugian (Juta Rp)')
plt.ylabel('Jenis Bencana')
for idx, val in enumerate(df_rugi['Kerugian_Juta']):
    plt.text(val + 20, idx, f'Rp {val:,.1f} Jt', va='center')
plt.xlim(0, 1850)
plt.tight_layout()
plt.show()
```

```python
# 7.3 Heatmap Korelasi Spearman
plt.figure(figsize=(8, 6))
corr_vars = ['Kejadian', 'Kecamatan_Terdampak', 'Total_Rumah_Rusak', 'Warga_Terdampak_Jiwa', 'Jumlah_Kerugian']
sns.heatmap(df[corr_vars].corr(method='spearman'), annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Matriks Korelasi Spearman Antar Variabel Utama', fontweight='bold')
plt.tight_layout()
plt.show()

# 8.1 Deteksi Outliers dengan Metode IQR
for col in ['Kejadian', 'Warga_Terdampak_Jiwa', 'Jumlah_Kerugian']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[df[col] > upper_bound][['Jenis_Bencana', col]]
    print(f'Outliers pada {col} (Batas Atas = {upper_bound:,.2f}):')
    print(outliers.to_string(index=False), '\n')
```
